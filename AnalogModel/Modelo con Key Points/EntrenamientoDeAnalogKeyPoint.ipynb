```python
import os
from ultralytics import YOLO

# 1. Ajuste de Directorio (CRUCIAL EN KAGGLE)
# Forzamos que el script se ejecute en la carpeta donde tenemos permisos de escritura
os.chdir('/kaggle/working')

# 2. Inicialización del Modelo
print("Cargando el modelo YOLOv8n-pose...")
model = YOLO('yolov8n-pose.pt') 

# 3. Configuración y Ejecución del Entrenamiento
print("Iniciando el entrenamiento...")
resultados = model.train(
    data=f"{dataset.location}/data.yaml", # Apunta automáticamente al dataset descargado
    epochs=50,           # Límite máximo de épocas
    patience=5,          # Early stopping: se detiene si el modelo no mejora durante 5 épocas seguidas
    imgsz=640,            # Resolución estándar y segura para las imágenes
    batch=16,             # Imágenes por lote
    device=0,             # Kaggle reconocerá la GPU en el índice 0
    project="/kaggle/working/Lector_Manometro", # Ruta absoluta de Kaggle
    name="version_1"      # Subcarpeta donde se guardará este entrenamiento específico
)

print("¡Entrenamiento finalizado!")
print("Descarga tus pesos finales en: /kaggle/working/Lector_Manometro/version_1/weights/best.pt")
```
