```python
import os
from roboflow import Roboflow
from ultralytics import YOLO

# 1. Autenticación y descarga
rf = Roboflow(api_key="jNIbs6DBulk39NECxeAT")
project = rf.workspace("luizfelipepabs-workspace-goczc").project("gauge-right-dj-i9sw8")
version = project.version(1)
dataset = version.download("yolov8")

# 2. Carga del modelo base
modelo = YOLO("yolov8n-seg.pt")

print("\nIniciando el entrenamiento del modelo de segmentación...")
resultados = modelo.train(
    data=f"{dataset.location}/data.yaml", 
    epochs=100,        
    imgsz=640,         
    batch=16,          
    patience=20,       
    device=0           
)

# 3. Exportación a la Jetson Nano (ONNX)
mejor_modelo_path = "runs/segment/train/weights/best.pt"

print(f"\nExportando el modelo {mejor_modelo_path} al formato ONNX...")
modelo_otimizado = YOLO(mejor_modelo_path)

# Exportación fijada en Opset 11 para garantizar la compatibilidad con TensorRT 8.2 de la Jetson
modelo_otimizado.export(
    format="onnx",
    imgsz=640,
    opset=11
)

arquivo_onnx = "runs/segment/train/weights/best.onnx"

if os.path.exists(arquivo_onnx):
    print(f"\n[ÉXITO] El modelo final está guardado en: {arquivo_onnx}")
    print("Descargue este archivo best.onnx desde el menú lateral izquierdo de Kaggle (carpeta 'runs/...') y transfiéralo a la Jetson Nano.")
else:
    print("\n[ERROR] El archivo ONNX no fue generado.")
```
